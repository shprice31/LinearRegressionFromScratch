```python
def _validate_normal_equation_inputs(X: pd.DataFrame, y: pd.Series):
    
    #X has the same number of rows as y
    if len(X) != len(y):
        raise ValueError("X and y have different lengths")

    #Neither x nor y has missing values
    if X.isna().any(axis = None):
        raise ValueError("X has at least 1 missing value")
    
    if y.isna().any():
        raise ValueError("y has at least 1 missing value")

    #If x has a single columm, it has to have at least one non-zero value
    if X.shape[1] == 1:
        if not X.any().item():
            raise ValueError("If X has a single column, it must have at least 1 non-zero value")

def fit_using_normal_equation(X: pd.DataFrame, y: pd.Series) -> pd.DataFrame:
    
    _validate_normal_equation_inputs(X, y)

    #reshape
    X = np.array(X) #type: ignore
    y = np.array(y) #type: ignore

    X_t = X.T
    beta_hat = np.linalg.solve(X_t @ X, X_t @ y)

    return pd.DataFrame(beta_hat)


def _validate_predict_using_lin_reg_inputs(X: pd.DataFrame, beta_hat: pd.DataFrame):

    #X has the same number of columns as beta_hat does rows
    if X.shape[1] != beta_hat.shape[0]:
        raise ValueError(f"X has {X.shape[1]} columns while beta_hat has {beta_hat.shape[0]} rows")

    #beta_hat isn't 0d
    if len(beta_hat.shape) == 1:
        raise ValueError("Beta_hat cannot be 0d")

    #beta_hat has 1 column
    if beta_hat.shape[1] != 1:
        raise ValueError(f"Beta_hat must have 1 column. It has {beta_hat.shape[1]}")
    

def predict_using_lin_reg(X: pd.DataFrame, beta_hat: pd.DataFrame):
    """
    Multiply coefficients by feature matrix to generate predictions
    """

    _validate_predict_using_lin_reg_inputs(X, beta_hat)

    print(X.shape)
    print(beta_hat.shape)

    print(type(X))
    print(type(beta_hat))

    preds = X @ beta_hat

    return preds


```
